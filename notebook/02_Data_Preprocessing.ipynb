```python
# ==========================================
# STEP 1 - Import Libraries
# ==========================================

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
```

```python
# ==========================================
# STEP 2 - Load Dataset
# ==========================================

df = pd.read_csv("../data/Customer360Insights.csv")

df.head()
```

```python
# ==========================================
# STEP 3 - Check Data Types
# ==========================================

df.info()
```

```python
# ==========================================
# STEP 4 - Convert Datetime
# ==========================================

datetime_columns = [
    "SessionStart",
    "CartAdditionTime",
    "OrderConfirmationTime",
    "SessionEnd"
]

for col in datetime_columns:
    df[col] = pd.to_datetime(df[col])

df.info()
```

```python
# ==========================================
# STEP 5 - Missing Values
# ==========================================

missing = pd.DataFrame({
    "Missing": df.isnull().sum(),
    "Percent": round(df.isnull().mean() * 100,2)
})

missing[missing["Missing"] > 0]
```

```python
# ==========================================
# STEP 6 - Session Features
# ==========================================

df["SessionHour"] = df["SessionStart"].dt.hour

df["SessionDay"] = df["SessionStart"].dt.day

df["SessionMonth"] = df["SessionStart"].dt.month

df["Weekday"] = df["SessionStart"].dt.day_name()
```

```python
# ==========================================
# STEP 7 - Total Amount
# ==========================================

df["TotalAmount"] = df["Price"] * df["Quantity"]

df[["Price","Quantity","TotalAmount"]].head()
```

```python
# ==========================================
# STEP 8 - Profit
# ==========================================

df["Profit"] = df["Price"] - df["Cost"]

df[["Cost","Price","Profit"]].head()
```

```python
# ==========================================
# STEP 9 - Check New Columns
# ==========================================

df.columns
```

```python
# ==========================================
# STEP 10 - Remove Unnecessary Columns
# ==========================================

drop_columns = [
    "CustomerID",
    "FullName"
]

df = df.drop(columns=drop_columns)

df.head()
```

```python
# ==========================================
# STEP 11 - Save Clean Dataset
# ==========================================

df.to_csv("../data/customer_preprocessed.csv",index=False)
```
