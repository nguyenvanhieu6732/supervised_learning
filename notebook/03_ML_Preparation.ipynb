```python
# ==========================================
# STEP 1 - Import Libraries
# ==========================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import OneHotEncoder

pd.set_option("display.max_columns", None)
```

```python
# ==========================================
# STEP 2 - Load Dataset
# ==========================================

df = pd.read_csv("../data/customer_preprocessed.csv")

df.head()
```

```python
# ==========================================
# STEP 3 - Dataset Overview
# ==========================================

print(df.shape)

df.info()
```

```python
# ==========================================
# STEP 4 - Select Target
# ==========================================

target = "OrderConfirmation"
```

```python
# ==========================================
# STEP 5 - Remove Data Leakage
# ==========================================

drop_columns = [

    "OrderConfirmationTime",

    "PaymentMethod",

    "OrderReturn",

    "ReturnReason"

]
```

```python
drop_columns += [

    "SessionStart",

    "CartAdditionTime",

    "SessionEnd"

]
```

```python
# ==========================================
# STEP 7 - Split Feature & Target
# ==========================================

X = df.drop(columns=drop_columns + [target])

y = df[target]
```

```python
X.head()
```

```python
X.info()
```

```python
# ==========================================
# STEP 9 - Find Categorical Columns
# ==========================================

categorical_columns = X.select_dtypes(include="object").columns

categorical_columns
```

```python
# ==========================================
# STEP 10 - One Hot Encoding
# ==========================================

encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

encoded = encoder.fit_transform(
    X[categorical_columns]
)
```

```python
encoded_df = pd.DataFrame(

    encoded,

    columns=encoder.get_feature_names_out(categorical_columns),

    index=X.index

)
```

```python
X = X.drop(columns=categorical_columns)

X = pd.concat(
    [X, encoded_df],
    axis=1
)
```

```python
# ==========================================
# STEP 13 - Convert Target
# ==========================================

y = y.astype(int)
```

```python
print(X.shape)

X.info()

X.head()

y.head()
```

```python
# ==========================================
# STEP 15 - Save Dataset
# ==========================================

X.to_csv("../data/X.csv", index=False)

y.to_csv("../data/y.csv", index=False)
```
